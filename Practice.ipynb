```python
import matplotlib.pyplot as plt
```

```python
height=[140,150,160,170,180,190,200,210,220,230]
weight=[40,45,50,55,60,65,70,75,80,85]
plt.scatter(height,weight)
plt.title("Height vs Weight")
plt.xlabel("height in (cm)")
plt.ylabel("Weight in (kg)")
plt.savefig("height_weight_plot.png")
plt.show()
```

```python
print("Chatbot: Hello! Type 'bye' to exit.")

while True:
    user_input = input("You: ").lower()

    if user_input == "hi":
        print("Chatbot: Hello")
    elif user_input == "good morning! how are you":
        print("Chatbot: Good Morning! I am fine.")
    elif user_input == "have a nice day":
        print("Chatbot: Thank you! You too.")
    elif user_input == "bye":
        print("Chatbot: Goodbye! Have a great day.")
        break
    else:
        print("Chatbot: Sorry, I don't understand.")
```
